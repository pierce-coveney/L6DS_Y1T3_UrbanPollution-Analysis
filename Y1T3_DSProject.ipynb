### Analysis and Modelling

Now that the scatteplots have provided an initial understanding of how these specific polution levels are trending over time in each city, lets try to understand which cities are experiencing an upward trend in their pollution figures and may face potentially dangerous levels of 'PM2_5_ug_m3' pollutants in the near future.

Trend is most commonly measured using Regression analysis. In this case, a linear regression model should be established to calculate the direction of trend for each city within the dataset.

```python
### Time series analysis with regression modelling requires a linear scale that represents time rather than a Timestamp date format.
### Start by converting the Timestamp column into the number of days since the earliest date. 
df['Days since start'] = (df['Timestamp'] - df['Timestamp'].min()).dt.days

### Begin the next step by creating an empty list to store results of the later regression model.
trend_results = []

### Since a regression analysis needs to be run for each City, create a loop through each city:
for city in df['City'].unique():

    ### In each version of the regression analysis on the dataframe 'city_df', the 'City' column is filtered to only present a single city per loop. 
    city_df = df[df['City'] == city]
    
    ### The final check, the regression should only be run if there are enough observations within the dataframe.
    if len(city_df) >= 20:
        
        regression = linregress(
            city_df['Days since start'],
            city_df['PM2_5_ug_m3']
        )
        
        slope = regression.slope
        p_value = regression.pvalue
        r_squared = regression.rvalue ** 2

        ### With the regression model parameters in place, a column titled 'trend' then describes if the trend is statistically significant. 
        if slope > 0 and p_value < 0.05:
            trend = 'Significant upward trend'
        elif slope < 0 and p_value < 0.05:
            trend = 'Significant downward trend'
        elif slope > 0:
            trend = 'Upward trend, not statistically significant'
        elif slope < 0:
            trend = 'Downward trend, not statistically significant'
        else:
            trend = 'No trend'

        ### The outputs of the regression model are finally presented by populating the empty list created at the start. 
        trend_results.append({
            'City': city,
            'Number of records': len(city_df),
            'Slope per day': slope,
            'P-value': p_value,
            'R-squared': r_squared,
            'Trend': trend
        })

### The looping regression model results are then converted to a dataframe for easier sorting and presentation.
trend_summary = pd.DataFrame(trend_results)

### Outputs are sorted so the greatest upward slopes appear first.
trend_summary = trend_summary.sort_values(
    by='Slope per day',
    ascending=False
)

trend_summary
```
